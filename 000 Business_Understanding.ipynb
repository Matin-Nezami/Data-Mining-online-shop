# سوال 1: مسئله اصلی پروژه رو با زبان ساده توضیح دهید

## چرا؟
پاسخ به این سؤال کمک می‌کند:
- مسئله اصلی فروشگاه را به‌درستی شناسایی کنیم.
- از اشتباه گرفتن مشکل کسب‌وکار با راه‌حل فنی جلوگیری کنیم.
- هدف و مسیر اصلی پروژه را مشخص کنیم.
- مطمئن شویم روش‌های داده‌کاوی که در مراحل بعد استفاده می‌کنیم، در راستای حل یک نیاز واقعی کسب‌وکار هستند.

در واقع، این سوال کمک می‌کند بدانیم قراره چه مشکلی را حل کنیم، نه اینکه صرفاً چه مدلی بسازیم.

## پاسخ:
این فروشگاه در طول فعالیت خود اطلاعات زیادی از سفارش‌ها و خریدهای مشتریان جمع‌آوری کرده است؛ اطلاعاتی مانند تعداد سفارش‌ها، تاریخ خرید، محصولات خریداری‌شده، تعداد کالاها و مبلغ هر سفارش. با این حال، صرفا داشتن این اطلاعات به معنای شناخت مشتریان نیست. فروشگاه باید بتواند این داده‌ها را تحلیل کند تا متوجه شود مشتریان چگونه خرید می‌کنند، چه الگوهای رفتاری دارند و چه تفاوت‌هایی بین آن‌ها وجود دارد.

برای مثال، ممکن است بعضی از مشتریان در طول زمان خریدهای مکرر و با مبالغ بالا داشته باشند، بعضی دیگر فقط در چند نوبت خرید کنند و گروهی نیز تعداد سفارش‌های زیادی با مبالغ پایین ثبت کنند. اگر فروشگاه نتواند این تفاوت‌ها را شناسایی کند، نمی‌تواند مشتریان خود را به‌درستی دسته‌بندی کند یا برای هر گروه برنامه‌ریزی مناسبی داشته باشد.

این موضوع باعث می‌شود تصمیم‌گیری درباره تبلیغات، پیشنهاد محصولات، برنامه‌های بازاریابی و نحوه ارتباط با مشتریان بیشتر بر اساس حدس و تجربه انجام شود، نه بر اساس تحلیل واقعی رفتار خرید آن‌ها.

در نتیجه، مسئله اصلی کسب‌وکار این است که با وجود در اختیار داشتن حجم زیادی از اطلاعات فروش، شناخت و دسته‌بندی دقیقی از رفتار خرید مشتریان ندارد و نمی‌تواند به‌صورت مؤثر از این اطلاعات برای تصمیم‌گیری‌های بهتر و بهبود عملکرد فروشگاه استفاده کند.

# سوال 2 : اگر این پروژه اجرا نشود، چه مشکلاتی برای بیمارستان ایجاد خواهد شد؟  

## چرا؟
در مجموع، اجرا نشدن این پروژه باعث می‌شود فروشگاه فرصت شناخت بهتر مشتریان، شناسایی الگوهای خرید و استفاده مؤثر از اطلاعات فروش را از دست بدهد. در نتیجه، تصمیم‌گیری‌های بازاریابی و تجاری آن با دقت کمتری انجام خواهد شد و امکان استفاده بهتر از ظرفیت داده‌های موجود کاهش پیدا می‌کند.

## پاسخ:
مهم‌ترین مشکلاتی که ممکن است برای فروشگاه ایجاد شود عبارت‌اند از:

1. شناخت ناکافی از مشتریان: فروشگاه نمی‌تواند به‌درستی تشخیص دهد مشتریان چه عادت‌ها و الگوهایی در خرید دارند و میزان خرید آن‌ها چگونه با یکدیگر متفاوت است.

2. ناتوانی در دسته‌بندی مشتریان: شناسایی مشتریانی که رفتار خرید مشابهی دارند دشوار خواهد بود و فروشگاه نمی‌تواند گروه‌های مختلف مشتریان را به‌درستی از یکدیگر تفکیک کند.

3. بازاریابی غیرهدفمند: بدون شناخت رفتار مشتریان، ارائه تبلیغات و پیشنهادهای متناسب با نیاز و علاقه گروه‌های مختلف دشوارتر می‌شود.

4. از دست رفتن فرصت‌های فروش: فروشگاه ممکن است نتواند مشتریان باارزش و گروه‌هایی را که ظرفیت خرید بیشتری دارند، به‌درستی شناسایی کند.

5. تصمیم‌گیری بر اساس حدس و تجربه: مدیران فروشگاه به‌جای استفاده از اطلاعات تحلیلی، بیشتر به روش‌های سنتی و تجربیات شخصی برای تصمیم‌گیری متکی خواهند بود.

6. استفاده نکردن از ظرفیت داده‌های موجود: حجم زیادی از اطلاعات خرید مشتریان جمع‌آوری شده است، اما بدون تحلیل مناسب، بخش مهمی از ارزش این اطلاعات ناشناخته باقی می‌ماند.

7. دشواری در پیش‌بینی رفتار مشتریان: فروشگاه ابزار مناسبی برای پیش‌بینی گروه رفتاری مشتریان جدید یا بررسی رفتار مشتریان بر اساس الگوهای شناسایی‌شده نخواهد داشت.

# سوال 3 : چه ذی نفعانی در این پروژه حضور دارند؟

## پاسخ:

در پروژه تحلیل رفتار خرید مشتریان فروشگاه آنلاین، ذی‌نفعان اصلی عبارت‌اند از:

1. مدیر فروشگاه (Store Manager): یکی از مهم‌ترین ذی‌نفعان پروژه است؛ زیرا از نتایج تحلیل رفتار مشتریان برای تصمیم‌گیری بهتر، برنامه‌ریزی فروش و بهبود عملکرد فروشگاه استفاده می‌کند.

2. تیم بازاریابی (Marketing Team): از اطلاعات به‌دست‌آمده برای شناخت گروه‌های مختلف مشتریان، طراحی تبلیغات هدفمند و ارائه پیشنهادهای متناسب با رفتار خرید آن‌ها استفاده می‌کند.

3. تیم فروش (Sales Team): می‌تواند با شناخت بهتر مشتریان و الگوهای خرید آن‌ها، فرصت‌های فروش را شناسایی کند و برنامه‌ریزی دقیق‌تری برای فروش محصولات داشته باشد.

4. تیم تحلیل داده (Data Analysis Team): مسئول بررسی، پاک‌سازی و آماده‌سازی داده‌ها، اجرای الگوریتم‌های خوشه‌بندی و طبقه‌بندی و تحلیل نتایج مدل‌ها است.

5. مشتریان فروشگاه (Customers): از ذی‌نفعان غیرمستقیم پروژه هستند؛ زیرا شناخت بهتر رفتار خرید آن‌ها می‌تواند به ارائه پیشنهادهای مرتبط‌تر و بهبود تجربه خریدشان کمک کند.

6. مدیران ارشد و صاحبان کسب‌وکار (Business Owners): از نتایج پروژه برای تصمیم‌گیری‌های کلان، تعیین استراتژی‌های تجاری و برنامه‌ریزی برای توسعه فروشگاه استفاده می‌کنند.

در مجموع، مدیران فروشگاه، تیم بازاریابی، تیم فروش، تیم تحلیل داده و مشتریان از مهم‌ترین ذی‌نفعان این پروژه هستند. هرکدام به شکل متفاوتی از نتایج تحلیل رفتار مشتریان بهره‌مند می‌شوند یا در اجرای پروژه نقش دارند.

### ۱. بررسی و پاک‌سازی اطلاعات مشتریان (Data Cleaning)

بررسی داده‌های فروشگاه آنلاین، شناسایی اطلاعات ناقص، داده‌های تکراری و مقادیر نامعتبر و آماده‌سازی داده‌ها برای انجام تحلیل‌های بعدی. هدف این است که اطلاعات مورد استفاده تا حد امکان دقیق و قابل اعتماد باشند.

### ۲. شناسایی الگوهای خرید مشتریان (Customer Behavior Analysis)

بررسی رفتار خرید مشتریان با استفاده از ویژگی‌هایی مانند تعداد سفارش‌ها، میزان خرید، مبلغ کل خرید و زمان خرید، به‌منظور شناخت تفاوت‌ها و شباهت‌های رفتاری آن‌ها.

### ۳. دسته‌بندی مشتریان بر اساس رفتار خرید (Customer Clustering)

استفاده از الگوریتم‌های خوشه‌بندی برای قرار دادن مشتریانی که الگوهای خرید مشابهی دارند در گروه‌های مشخص و ایجاد برچسب رفتاری برای هر گروه.

### ۴. ساخت مدل پیش‌بینی رفتار مشتریان (Customer Classification)

آموزش یک مدل طبقه‌بندی با استفاده از برچسب‌های تولیدشده در مرحله خوشه‌بندی، به‌طوری‌که مدل بتواند گروه رفتاری مشتریان را بر اساس ویژگی‌های خرید آن‌ها پیش‌بینی کند.

### ۵. آماده‌سازی مجموعه داده نهایی شامل ۵۰۰ رکورد

انتخاب و آماده‌سازی ۵۰۰ رکورد معتبر، تمیز و برچسب‌گذاری‌شده از مشتریان برای استفاده در تحلیل‌ها و مراحل مدل‌سازی پروژه.

---

### جمع‌بندی:

این اهداف فرعی در کنار یکدیگر کمک می‌کنند تا داده‌های خام فروشگاه به اطلاعات قابل استفاده تبدیل شوند، رفتار مشتریان بهتر شناخته شود و در نهایت یک مجموعه داده ۵۰۰ رکوردی همراه با برچسب‌های رفتاری و مدل پیش‌بینی‌کننده گروه مشتریان آماده شود.

# سؤال ۷: چه شاخص‌هایی نشان می‌دهد پروژه موفق بوده است؟

برای ارزیابی موفقیت این پروژه، باید شاخص‌هایی مشخص کنیم که بتوانیم با استفاده از آن‌ها میزان دستیابی به اهداف پروژه را به‌صورت عددی و قابل اندازه‌گیری بررسی کنیم.

در پروژه تحلیل رفتار خرید مشتریان فروشگاه آنلاین، شاخص‌های موفقیت (KPI) عبارت‌اند از:

---

### ۱. کیفیت داده‌های نهایی (Data Quality)

داده‌های نهایی باید تا حد امکان فاقد مقادیر ناقص، تکراری و نامعتبر باشند. موفقیت در این بخش با بررسی تعداد رکوردهای پاک‌سازی‌شده و میزان داده‌های ناقص و تکراری مشخص می‌شود.

### ۲. دستیابی به ۵۰۰ رکورد تمیز و برچسب‌گذاری‌شده

یکی از معیارهای اصلی موفقیت پروژه، آماده‌سازی دقیقاً ۵۰۰ رکورد معتبر از مشتریان است که پس از پاک‌سازی، دارای برچسب رفتاری مشخص باشند.

### ۳. کیفیت خوشه‌بندی مشتریان (Clustering Quality)

خوشه‌بندی باید بتواند مشتریان را بر اساس شباهت رفتار خرید در گروه‌های مشخص و قابل تفکیک قرار دهد. برای بررسی کیفیت این مرحله می‌توان از شاخص‌هایی مانند Silhouette Score استفاده کرد.

### ۴. دقت مدل طبقه‌بندی (Classification Performance)

مدل طبقه‌بندی باید بتواند گروه رفتاری مشتریان را با دقت قابل قبول پیش‌بینی کند. برای ارزیابی عملکرد آن می‌توان از معیارهایی مانند Accuracy، Precision، Recall و F1-Score استفاده کرد.

### ۵. قابلیت شناسایی گروه‌های رفتاری مشتریان

نتایج پروژه باید به ما کمک کنند گروه‌های مختلف مشتریان را شناسایی کنیم؛ برای مثال مشتریان با خریدهای مکرر، مشتریان با حجم خرید بالا و مشتریان با خریدهای محدودتر.

### ۶. قابلیت استفاده از نتایج در تصمیم‌گیری‌های تجاری

نتایج تحلیل باید برای مدیران فروشگاه قابل فهم و کاربردی باشند و بتوانند در برنامه‌ریزی بازاریابی، طراحی تبلیغات و ارائه پیشنهادهای مناسب به مشتریان مورد استفاده قرار بگیرند.

---

### جمع‌بندی

در مجموع، زمانی می‌توانیم این پروژه را موفق بدانیم که بتوانیم داده‌های فروشگاه را با کیفیت مناسب پاک‌سازی کنیم، ۵۰۰ رکورد معتبر و برچسب‌گذاری‌شده آماده کنیم، مشتریان را به‌درستی خوشه‌بندی کنیم و یک مدل طبقه‌بندی با عملکرد قابل قبول برای پیش‌بینی گروه رفتاری آن‌ها بسازیم.

### نکته مهم

برای شاخص‌هایی مانند Silhouette Score و Accuracy نباید از ابتدا عدد دلخواه تعیین کنیم. ابتدا مدل‌ها را اجرا و نتایج را بررسی می‌کنیم؛ سپس با توجه به کیفیت داده‌ها و عملکرد مدل، درباره قابل قبول بودن نتایج تصمیم می‌گیریم.

# سؤال 9 : موفقیت این پروژه از دید مدیریت فروشگاه چگونه اندازه‌گیری می‌شود؟

مهم‌ترین معیارهای سنجش موفقیت پروژه از دید مدیر فروشگاه عبارت‌اند از:

1. **افزایش میزان فروش:** بررسی اینکه آیا شناخت بهتر رفتار مشتریان و ارائه پیشنهادهای متناسب با نیاز آن‌ها می‌تواند باعث افزایش مبلغ فروش شود یا خیر.

2. **افزایش خرید مجدد مشتریان:** بررسی اینکه آیا با شناخت الگوهای خرید مشتریان می‌توان آن‌ها را به خریدهای مجدد تشویق کرد و میزان وفاداری آن‌ها را افزایش داد.

3. **شناسایی مشتریان باارزش:** بررسی اینکه آیا فروشگاه توانسته است مشتریانی را که خریدهای بیشتر یا مبالغ بالاتری دارند شناسایی کند و برای حفظ آن‌ها برنامه‌ریزی مناسبی داشته باشد.

4. **بهبود عملکرد بازاریابی:** بررسی اینکه آیا اطلاعات به‌دست‌آمده از پروژه به فروشگاه کمک کرده است تبلیغات و پیشنهادهای هدفمندتری برای گروه‌های مختلف مشتریان طراحی کند.

5. **کاهش هزینه‌های بازاریابی:** بررسی اینکه آیا با شناخت بهتر مشتریان می‌توان از تبلیغات غیرهدفمند و هزینه‌های اضافی جلوگیری کرد و بودجه بازاریابی را مؤثرتر مصرف کرد.

6. **بهبود تصمیم‌گیری مدیریتی:** بررسی اینکه آیا مدیر فروشگاه می‌تواند با استفاده از گزارش‌های تحلیلی و شناخت گروه‌های مختلف مشتریان، تصمیم‌های دقیق‌تر و آگاهانه‌تری درباره فروش، محصولات و برنامه‌های بازاریابی بگیرد.

7. **افزایش رضایت مشتریان:** بررسی اینکه آیا شناخت بهتر نیازها و رفتار خرید مشتریان باعث ارائه پیشنهادهای مرتبط‌تر و بهبود تجربه خرید آن‌ها شده است یا خیر.

در واقع، مهم‌ترین معیار مدیر فروشگاه این است که اجرای پروژه چه ارزش اقتصادی و تجاری برای فروشگاه ایجاد کرده است.

اطلاعاتی مانند محصولاتی که مشتری مشاهده کرده، مدت‌زمان حضور در سایت، صفحاتی که بازدید کرده و محصولاتی که به سبد خرید اضافه کرده اما خریداری نکرده است، ممکن است در سیستم فروشگاه ثبت نشده باشند یا دسترسی به آن‌ها دشوار باشد.

### ۳. اطلاعات مربوط به مشتریان جدید و ناشناس

برخی افراد بدون ایجاد حساب کاربری از فروشگاه خرید می‌کنند یا فقط محصولات را مشاهده می‌کنند. در نتیجه، شناسایی و ارتباط دادن فعالیت‌های مختلف آن‌ها به یک مشتری مشخص دشوار خواهد بود.

### ۴. اطلاعات مربوط به علت لغو سفارش‌ها

ممکن است اطلاعاتی درباره دلیل لغو سفارش یا انصراف مشتری از خرید در دسترس نباشد. برای مثال، مشخص نباشد مشتری به دلیل قیمت بالا، هزینه ارسال، تأخیر در تحویل یا تغییر نظر خود از خرید منصرف شده است.

### ۵. اطلاعات مربوط به رضایت و نظرات مشتریان

جمع‌آوری اطلاعاتی مانند میزان رضایت مشتری، دلیل نارضایتی، تجربه خرید و نظر واقعی او درباره محصولات دشوار است؛ زیرا همه مشتریان در نظرسنجی‌ها شرکت نمی‌کنند یا نظر خود را ثبت نمی‌کنند.

### ۶. اطلاعات مربوط به خرید مشتریان از فروشگاه‌های دیگر

برای شناخت بهتر رفتار مشتریان، دانستن اینکه آن‌ها از چه فروشگاه‌های دیگری خرید می‌کنند یا چه محصولاتی را از رقبا تهیه می‌کنند می‌تواند مفید باشد؛ اما دسترسی به این اطلاعات معمولاً بسیار دشوار است، زیرا در اختیار فروشگاه‌های دیگر قرار دارد.

## سؤال ۱۶: هدف داده‌کاوی این پروژه چیست؟

هدف داده‌کاوی این است که با استفاده از اطلاعات ثبت‌شده در تراکنش‌های فروشگاه، رفتار خرید مشتریان را تحلیل کنیم، مشتریانی را که الگوهای خرید مشابهی دارند شناسایی کنیم و آن‌ها را در گروه‌های رفتاری مختلف قرار دهیم.

برای رسیدن به این هدف، پروژه در دو مرحله انجام می‌شود:

### ۱. خوشه‌بندی مشتریان (Clustering):

با استفاده از الگوریتم K-Means و ویژگی‌هایی مانند تعداد سفارش‌ها، مجموع تعداد کالاهای خریداری‌شده، مبلغ کل خرید و فاصله زمانی از آخرین خرید، مشتریان را بر اساس شباهت رفتار خریدشان گروه‌بندی می‌کنیم. سپس برای هر گروه یک برچسب رفتاری مشخص در نظر می‌گیریم.

### ۲. طبقه‌بندی مشتریان (Classification):

پس از ایجاد گروه‌های رفتاری، از برچسب‌های حاصل از خوشه‌بندی برای آموزش مدل‌های طبقه‌بندی استفاده می‌کنیم تا مدل بتواند گروه رفتاری مشتریان را بر اساس اطلاعات خرید آن‌ها پیش‌بینی کند.

در نهایت، هدف این است که یک مجموعه داده شامل ۵۰۰ مشتری معتبر، پاک‌سازی‌شده و برچسب‌گذاری‌شده آماده کنیم که اطلاعات رفتاری و گروه پیش‌بینی‌شده هر مشتری در آن مشخص باشد.

## سؤال 17: تفاوت بین Business Goal و Data Mining Goal چیست؟

| Business Goal | Data Mining Goal |
|---|---|
| تمرکز بر نیاز و منافع فروشگاه دارد. | تمرکز بر تحلیل داده‌ها و ساخت مدل دارد. |
| به دنبال بهبود عملکرد کسب‌وکار است. | به دنبال کشف الگوهای پنهان در داده‌هاست. |
| شناخت بهتر مشتریان و افزایش فروش را دنبال می‌کند. | مشتریان را خوشه‌بندی و گروه رفتاری آن‌ها را پیش‌بینی می‌کند. |
| از دید مدیر فروشگاه تعریف می‌شود. | از دید تیم تحلیل داده تعریف می‌شود. |

### یک مثال ساده از پروژه خودمان

فرض کنیم مدیر فروشگاه می‌خواهد بداند کدام مشتریان ارزش بیشتری برای فروشگاه دارند تا بتواند برای آن‌ها برنامه‌های بازاریابی مناسبی طراحی کند.

- **Business Goal:** شناسایی بهتر مشتریان باارزش و بهبود استراتژی‌های بازاریابی برای افزایش فروش.
- **Data Mining Goal:** تحلیل اطلاعات خرید مشتریان، استخراج ویژگی‌های رفتاری، خوشه‌بندی مشتریان و آموزش مدل طبقه‌بندی برای پیش‌بینی گروه رفتاری آن‌ها.

---

### هدف از این سوال:
هدف این سؤال این است که یاد بگیریم بین خواسته و نیاز واقعی کسب‌وکار و راهکار فنی داده‌کاوی تفاوت قائل شویم.

به زبان ساده، Business Goal مقصد پروژه است و Data Mining Goal مسیر تحلیلی رسیدن به آن مقصد است.

## سؤال 20 : چه معیارهایی برای ارزیابی مدل اهمیت دارند؟

### ۱. معیارهای ارزیابی Clustering (خوشه‌بندی):
- **Silhouette Score:** میزان شباهت اعضای هر خوشه به یکدیگر و میزان تفاوت آن‌ها با اعضای خوشه‌های دیگر را نشان می‌دهد. هرچه مقدار آن به ۱ نزدیک‌تر باشد، خوشه‌بندی بهتر است.
- **Davies-Bouldin Index:** میزان شباهت و هم‌پوشانی خوشه‌ها را بررسی می‌کند. هرچه مقدار آن کمتر باشد، نتیجه بهتر است.

### ۲. معیارهای ارزیابی Classification (طبقه‌بندی):
- **Accuracy:** درصد پیش‌بینی‌های درست مدل از کل پیش‌بینی‌ها.
- **Precision:** از بین مشتریانی که مدل در یک گروه قرار داده، چه تعداد واقعاً به همان گروه تعلق دارند.
- **Recall:** از بین مشتریانی که واقعاً متعلق به یک گروه هستند، مدل چه تعداد را درست شناسایی کرده است.
- **F1-Score:** میانگینی از Precision و Recall است و تعادل بین این دو معیار را نشان می‌دهد.
- **Confusion Matrix:** جدولی برای بررسی تعداد پیش‌بینی‌های درست و اشتباه مدل در هر گروه رفتاری است.

## سؤال 21 : Accuracy چیست و آیا برای این پروژه کافی است؟

Accuracy (دقت کلی) نشان می‌دهد مدل از بین تمام پیش‌بینی‌هایی که انجام داده، چند درصد را درست پیش‌بینی کرده است.

برای مثال، اگر مدل گروه رفتاری ۱۰۰ مشتری را پیش‌بینی کند و ۸۵ پیش‌بینی درست باشد، Accuracy برابر با ۸۵ درصد خواهد بود.

اما Accuracy به‌تنهایی برای پروژه ما کافی نیست. چون ممکن است تعداد مشتریان در گروه‌های رفتاری مختلف برابر نباشد. در این شرایط، مدل ممکن است در تشخیص یک گروه عملکرد خوبی داشته باشد، اما Accuracy کلی همچنان بالا باشد.

به همین دلیل، علاوه بر Accuracy باید معیارهای زیر را هم بررسی کنیم:

- **Precision:** میزان درستی پیش‌بینی‌های مدل برای هر گروه.
- **Recall:** میزان توانایی مدل در شناسایی درست اعضای هر گروه.
- **F1-Score:** تعادل بین Precision و Recall.
- **Confusion Matrix:** بررسی جزئیات پیش‌بینی‌های درست و اشتباه.

### نتیجه:
در پروژه ما Accuracy معیار مهمی است، اما به‌تنهایی کافی نیست و باید در کنار Precision، Recall و F1-Score بررسی شود تا ارزیابی دقیق‌تری از عملکرد مدل طبقه‌بندی داشته باشیم.

## سؤال 22 : موفقیت مدل را چگونه تعریف می‌کنید؟

برای ارزیابی موفقیت مدل، معیارهای زیر را در نظر می‌گیریم:

1. **کیفیت خوشه‌بندی:** گروه‌های رفتاری مشخص و قابل تفکیکی ایجاد شوند و Silhouette Score در سطح قابل قبولی باشد.
2. **دقت پیش‌بینی:** مدل Classification بتواند گروه رفتاری مشتریان را با عملکرد مناسب پیش‌بینی کند.
3. **عملکرد متعادل:** معیارهای Precision، Recall و F1-Score وضعیت مناسبی داشته باشند.
4. **قابلیت اعتماد:** مدل روی داده‌های آزمایشی نیز عملکرد قابل قبولی داشته باشد، نه فقط داده‌های آموزشی.
5. **کاربردپذیری:** نتایج مدل قابل فهم باشند و بتوان از آن‌ها برای شناخت بهتر مشتریان و برنامه‌ریزی بازاریابی استفاده کرد.
6. **تکمیل خروجی پروژه:** در نهایت ۵۰۰ رکورد معتبر، تمیز و برچسب‌گذاری‌شده آماده شود.

## سؤال 23 : Precision چیست؟

Precision (دقت پیش‌بینی مثبت) معیاری است که نشان می‌دهد از بین تمام مواردی که مدل در یک گروه قرار داده، چه تعداد واقعاً به همان گروه تعلق دارند.

به زبان ساده، Precision میزان قابل اعتماد بودن پیش‌بینی‌های مثبت مدل را نشان می‌دهد.

### فرمول Precision:

\[
\text{Precision} = \frac{TP}{TP + FP}
\]

- **TP (True Positive):** مواردی که مدل درست تشخیص داده است.
- **FP (False Positive):** مواردی که مدل اشتباهاً مثبت تشخیص داده است.

### مثال در پروژه خودمان:

فرض کنیم مدل، ۱۰۰ مشتری را در گروه «مشتریان با خرید بالا» قرار داده است؛ اما پس از بررسی مشخص می‌شود فقط ۸۰ نفر واقعاً به این گروه تعلق دارند.

در این صورت:

\[
\text{Precision} = \frac{80}{100} = 0.8 = 80\%
\]

یعنی ۸۰ درصد پیش‌بینی‌های مثبت مدل درست بوده‌اند.

پس هرچه Precision بالاتر باشد، می‌توانیم اعتماد بیشتری به پیش‌بینی‌های مثبت مدل داشته باشیم و احتمال قرار گرفتن اشتباه مشتریان در یک گروه رفتاری کمتر می‌شود.

هدف داده‌کاوی این پروژه، تحلیل داده‌های تراکنشی فروشگاه آنلاین با استفاده از دیتاست Online Retail به‌منظور شناسایی الگوهای خرید و رفتار مشتریان است. در این پروژه، ابتدا مشتریان بر اساس ویژگی‌های خرید و میزان شباهت رفتاری با استفاده از روش خوشه‌بندی (Clustering) گروه‌بندی می‌شوند. سپس با استفاده از برچسب‌های ایجادشده، یک مدل طبقه‌بندی (Classification) آموزش داده می‌شود تا بتواند گروه رفتاری مشتریان را پیش‌بینی کند.

## سؤال 25 : رابطه بین هدف داده‌کاوی و هدف کسب‌وکار را توضیح دهید.

### در پروژه ما:

### ۱. هدف کسب‌وکار (Business Goal):

شناخت بهتر مشتریان، شناسایی الگوهای خرید، دسته‌بندی مشتریان بر اساس رفتار خرید و کمک به بهبود تصمیم‌گیری‌های بازاریابی و افزایش فروش.

### ۲. هدف داده‌کاوی (Data Mining Goal):

تحلیل داده‌های فروشگاه آنلاین، استخراج ویژگی‌های مرتبط با خرید مشتریان، اجرای خوشه‌بندی (Clustering) برای شناسایی گروه‌های رفتاری و استفاده از طبقه‌بندی (Classification) برای پیش‌بینی گروه رفتاری مشتریان.

### رابطه بین این دو هدف:

هدف داده‌کاوی در واقع ابزاری برای رسیدن به هدف کسب‌وکار است. یعنی مدل‌های داده‌کاوی باید اطلاعاتی تولید کنند که به فروشگاه کمک کند مشتریان خود را بهتر بشناسد و تصمیم‌های دقیق‌تری در زمینه فروش و بازاریابی بگیرد.

**مثال:** اگر هدف کسب‌وکار شناسایی مشتریان باارزش باشد، هدف داده‌کاوی می‌تواند تحلیل میزان خرید، تعداد سفارش‌ها و زمان آخرین خرید مشتریان برای شناسایی گروه‌های رفتاری مختلف باشد.
